```python
import cv2
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

# Load the image
image = cv2.imread("pho.jpg")

# Check if image was loaded successfully
if image is None:
    print("Error: Could not load image. Please ensure 'pho.jpg' is in the correct directory.")
    exit()

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Reshape image into a list of pixels
pixels = image.reshape((-1, 3))

# Apply KMeans to find two dominant colors
kmeans = KMeans(n_clusters=7, random_state=0) # Changed n_clusters to 2 for two dominant colors
kmeans.fit(pixels)

# Get dominant colors
dominant_colors = kmeans.cluster_centers_

print("Detected Dominant Colors:")
color_patches = []
hex_codes = []

for i, color in enumerate(dominant_colors):
    r, g, b = color.astype(int)
    hex_code = '#{:02x}{:02x}{:02x}'.format(r, g, b)
    print(f"Color {i+1} RGB: ({r}, {g}, {b})")
    print(f"Color {i+1} HEX: {hex_code}")
    hex_codes.append(hex_code)

    color_patch = np.zeros((100, 200, 3), dtype="uint8")
    color_patch[:] = [b, g, r] # Changed from [r, g, b] to [b, g, r] for BGR format compatible with cv2_imshow
    color_patches.append(color_patch)

# Display both detected color patches
try:
    # Concatenate color patches horizontally for display
    combined_patch = np.concatenate(color_patches, axis=1)
    cv2_imshow(combined_patch)
except NameError:
    print("Warning: cv2_imshow not found, falling back to matplotlib.")
    fig, axes = plt.subplots(1, 2, figsize=(8, 4))
    for i, patch in enumerate(color_patches):
        # For matplotlib, the color patch should be RGB, so we revert to [r,g,b]
        axes[i].imshow(patch[:, :, ::-1]) # Display as RGB for matplotlib if using BGR array
        axes[i].set_title(f"Color {i+1} ({hex_codes[i]})")
        axes[i].axis('off')
    plt.tight_layout()
    plt.show()
except Exception as e:
    print(f"Warning: Could not display images. Error: {e}")
    fig, axes = plt.subplots(1, 2, figsize=(8, 4))
    for i, patch in enumerate(color_patches):
        # For matplotlib, the color patch should be RGB, so we revert to [r,g,b]
        axes[i].imshow(patch[:, :, ::-1]) # Display as RGB for matplotlib if using BGR array
        axes[i].set_title(f"Color {i+1} ({hex_codes[i]})")
        axes[i].axis('off')
    plt.tight_layout()
    plt.show()
```
